# Experiment 23: Bayesian Network Construction and Visualization

**Category**: CO4 – Linear Factor Models and Probabilistic Graphical Models  
**Objective**: Construct a Bayesian directed acyclic graphical model, define local conditional probability tables, and visualize dependency flows.


### Task 1: Define Directed Acyclic Graph (DAG) Structure and Conditional Probabilities


```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Classical Student Academic Evaluation Bayesian Network
# Nodes: Course_Difficulty (D), Intelligence (I), Exam_Grade (G), SAT_Score (S), Recommendation_Letter (L)

edges = [
    ('Difficulty', 'Grade'),
    ('Intelligence', 'Grade'),
    ('Intelligence', 'SAT'),
    ('Grade', 'Letter')
]

G_bayes = nx.DiGraph(edges)
print("Graph Nodes:", list(G_bayes.nodes))
print("Graph Edges (Direct Dependencies):", list(G_bayes.edges))
```

### Task 2: Define Conditional Probability Distributions (CPDs)


```python
# Conditional Probability Tables:
# P(D): [Easy: 0.6, Hard: 0.4]
cpd_D = {'Easy': 0.6, 'Hard': 0.4}

# P(I): [Low: 0.7, High: 0.3]
cpd_I = {'Low': 0.7, 'High': 0.3}

# P(S | I): P(SAT=High | I)
cpd_S_given_I = {'Low': 0.15, 'High': 0.85}

# P(L | Grade): P(Letter=Strong | Grade)
cpd_L_given_G = {'A': 0.90, 'B': 0.50, 'C': 0.10}

print("Conditional Probability Distributions Defined:")
print("P(Difficulty):", cpd_D)
print("P(Intelligence):", cpd_I)
print("P(Letter=Strong | Grade):", cpd_L_given_G)
```

### Task 3: Render Bayesian Network Directed Graph Layout


```python
plt.figure(figsize=(8, 6))
pos = {
    'Difficulty': (0, 2),
    'Intelligence': (2, 2),
    'Grade': (1, 1),
    'SAT': (3, 1),
    'Letter': (1, 0)
}

node_colors = ['#457b9d', '#2a9d8f', '#e76f51', '#f4a261', '#9b5de5']
nx.draw_networkx_nodes(G_bayes, pos, node_size=2800, node_color=node_colors, edgecolors='k', linewidths=1.8)
nx.draw_networkx_labels(G_bayes, pos, font_size=10, font_weight='bold', font_color='white')
nx.draw_networkx_edges(G_bayes, pos, edge_color='#1d3557', arrowsize=25, arrowstyle='->', width=2.0, min_source_margin=18, min_target_margin=18)

plt.title("Bayesian Directed Acyclic Graph (Student Performance Network)", fontweight='bold', fontsize=12)
plt.axis('off')
plt.tight_layout()
plt.show()
```
